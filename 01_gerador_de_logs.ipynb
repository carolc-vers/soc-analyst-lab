```python
import random
from datetime import datetime, timedelta

#Dados de usuários
usuarios = ["rita lee", "bob marley", "lady gaga", "fernanda dumbra", "leticia collin"]
ips_normais = ["157.46.141.144", "35.167.135.0", "53.147.161.90", "55.37.86.40", "117.6.59.195"]

horario_atual = datetime(2026, 10, 19, 8, 0, 0)
logs = []

#Gerador dos logs de acesso
for i in range(10):
  usuario_sorteado = random.choice(usuarios)
  ip_sorteado = random.choice(ips_normais)
  horario_atual = horario_atual + timedelta(seconds=random.randint(1, 60))

  sorteio = random.randint(1, 100)

  if sorteio <= 90:
    status = "SUCESSO"
  else:
    status = "FALHA"

  linha = f"{horario_atual.strftime("%Y-%m-%d %H:%M:%S")},{usuario_sorteado},{ip_sorteado},{status}"
  logs.append(linha)

print(logs[0].split(","))


with open("logs_autenticacao.csv", "w") as arquivo:
  arquivo.write("horario,usuario,ip,status\n")
  for linha in logs:
    arquivo.write(linha + "\n")

with open("logs_autenticacao.csv") as arquivo:
  print(arquivo.read())

```
